```python

squares = {num: num**2 for num in range(1, 11)}

even_squares = {key: value for key, value in squares.items() if key % 2 == 0}

print(even_squares)
```

```python
squares = {num: num**2 for num in range(1, 6)}

swapped_dict = {value: key for key, value in squares.items()}

print(swapped_dict)
```

```python
from collections import defaultdict

my_dict = defaultdict(list)

my_dict['fruits'].append('apple')
my_dict['fruits'].append('banana')
my_dict['vegetables'].append('carrot')
my_dict['vegetables'].append('spinach')

print(dict(my_dict))
```

```python
def count_characters(text):
    char_count = {}

    for char in text:
        char_count[char] = char_count.get(char, 0) + 1

    return char_count


result = count_characters("hello world")

print(result)
```
